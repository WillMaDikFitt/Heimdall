# Calibration EDA

Run `python scripts/run_simulation.py` and `python scripts/run_calibration.py` first so SQLite has data.

This notebook inspects raw vs calibrated vs Open-Meteo reference for the configured Indian site.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))

from src.config import load_config
from src.ingestion.store import load_calibrated, load_latest_coefficients, load_readings, load_reference

cfg = load_config()
readings = load_readings(cfg["station"]["id"])
reference = load_reference()
calibrated = load_calibrated()
coeffs = load_latest_coefficients()

print(cfg["station"])
print("readings", readings.shape, "reference", reference.shape, "calibrated", calibrated.shape)
coeffs
```

```python
channel = "temperature_c"
subset = calibrated.loc[calibrated["channel"] == channel].copy()
subset[["raw", "calibrated", "reference_value"]].describe()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(subset["timestamp"], subset["raw"], label="raw", alpha=0.6)
ax.plot(subset["timestamp"], subset["calibrated"], label="calibrated")
ax.plot(subset["timestamp"], subset["reference_value"], label="Open-Meteo")
ax.set_title(channel)
ax.legend()
fig.autofmt_xdate()
plt.show()
```
